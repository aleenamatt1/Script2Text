```python
import cv2
import numpy as np
from matplotlib import pyplot as plt

# 1. Load handwritten image
img = cv2.imread("/Users/aaronbaggot/Desktop/Script2Text/data/samples/Handwritten_2025-09-18_110153_2.png")
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# 2. Preprocess: blur and threshold
blurred = cv2.GaussianBlur(gray, (5,5), 0)
_, binary = cv2.threshold(blurred, 0, 255, cv2.THRESH_BINARY_INV + cv2.THRESH_OTSU)

# 3. Morphological cleanup
kernel = cv2.getStructuringElement(cv2.MORPH_RECT, (3,3))
cleaned = cv2.morphologyEx(binary, cv2.MORPH_CLOSE, kernel, iterations=2)

# 4. Find contours (characters)
contours, _ = cv2.findContours(cleaned, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

# 5. Draw bounding boxes around detected characters
result = cv2.cvtColor(gray, cv2.COLOR_GRAY2BGR)
char_count = 0
for contour in contours:
    x, y, w, h = cv2.boundingRect(contour)
    if w > 5 and h > 5:  # Filter small noise
        cv2.rectangle(result, (x, y), (x+w, y+h), (0, 255, 0), 2)
        char_count += 1

# 6. Display results
fig, axes = plt.subplots(2, 2, figsize=(12, 10))

axes[0, 0].imshow(gray, cmap='gray')
axes[0, 0].set_title("Original Image")
axes[0, 0].axis('off')

axes[0, 1].imshow(blurred, cmap='gray')
axes[0, 1].set_title("Blurred Image")
axes[0, 1].axis('off')

axes[1, 0].imshow(binary, cmap='gray')
axes[1, 0].set_title("Binary (Threshold)")
axes[1, 0].axis('off')

axes[1, 1].imshow(cv2.cvtColor(result, cv2.COLOR_BGR2RGB))
axes[1, 1].set_title(f"Detected Characters: {char_count}")
axes[1, 1].axis('off')

plt.tight_layout()
plt.show()

print(f"✓ Total characters detected: {char_count}")
```
